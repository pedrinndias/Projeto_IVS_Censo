```python
# Importa a biblioteca pandas para manipulação de dados
import pandas as pd
# Importa a biblioteca sqlite3 para trabalhar com bancos de dados SQLite
import sqlite3
# Importa a biblioteca os para operações com arquivos e diretórios
import os

# Exibe mensagem inicial de auditoria
print("Iniciando a Auditoria de Integridade dos Arquivos...")

# Define o caminho para a pasta de banco de dados
caminho_bd = '../banco_de_dados/'
# Define o caminho do arquivo SQLite
arquivo_sqlite = caminho_bd + 'Banco_IVS_Essencial.db'
# Define o caminho do arquivo Excel
arquivo_excel = caminho_bd + 'Base_IVS_Essencial.xlsx'

# Verifica se o arquivo SQLite existe
if os.path.exists(arquivo_sqlite):
    # Calcula o tamanho do arquivo em MB
    tamanho_mb = os.path.getsize(arquivo_sqlite) / (1024 * 1024)
    # Exibe mensagem de sucesso com o tamanho do arquivo
    print("OK: Banco SQLite encontrado (" + str(round(tamanho_mb, 2)) + " MB)")
else:
    # Exibe mensagem de erro caso o arquivo não seja encontrado
    print("ERRO: Banco SQLite não encontrado!")

# Verifica se o arquivo Excel existe
if os.path.exists(arquivo_excel):
    # Calcula o tamanho do arquivo em MB
    tamanho_mb = os.path.getsize(arquivo_excel) / (1024 * 1024)
    # Exibe mensagem de sucesso com o tamanho do arquivo
    print("OK: Ficheiro Excel encontrado (" + str(round(tamanho_mb, 2)) + " MB)")
else:
    # Exibe mensagem de erro caso o arquivo não seja encontrado
    print("ERRO: Ficheiro Excel não encontrado!")

# Inicia auditoria no banco SQLite
print("\n--- Auditoria no Banco SQLite ---")
try:
    # Abre conexão com o banco SQLite
    conexao = sqlite3.connect(arquivo_sqlite)
    cursor = conexao.cursor()
    
    # Conta o número de linhas na tabela base_ivs_bruta
    cursor.execute("SELECT COUNT(*) FROM base_ivs_bruta")
    linhas_sql = cursor.fetchone()[0]
    print("Linhas presentes: " + str(linhas_sql) + " (Esperado: 468099)")
    
    # Conta setores com código nulo ou vazio
    cursor.execute("SELECT COUNT(*) FROM base_ivs_bruta WHERE CD_SETOR IS NULL OR CD_SETOR = ''")
    nulos_sql = cursor.fetchone()[0]
    print("Setores com código vazio/nulo: " + str(nulos_sql) + " (Esperado: 0)")
    
    # Conta setores duplicados
    cursor.execute("SELECT COUNT(*) - COUNT(DISTINCT CD_SETOR) FROM base_ivs_bruta")
    duplicatas_sql = cursor.fetchone()[0]
    print("Setores duplicados: " + str(duplicatas_sql) + " (Esperado: 0)")
    
    # Verifica o número de colunas presentes
    df_head = pd.read_sql("SELECT * FROM base_ivs_bruta LIMIT 0", conexao)
    colunas_sql = len(df_head.columns)
    print("Colunas presentes: " + str(colunas_sql) + " (Esperado: 31)")
    
    # Fecha a conexão com o banco
    conexao.close()
except Exception as e:
    # Exibe mensagem de erro caso não consiga ler o banco
    print("Erro ao ler o SQLite: " + str(e))

# Inicia auditoria no arquivo Excel
print("\n--- Auditoria no Ficheiro Excel ---")
# Informa que a leitura do Excel pode demorar
print("Lendo o Excel para conferencia (isto pode demorar um minuto)...")
try:
    # Lê as colunas CD_SETOR, v0001 e V00001 do Excel
    df_excel = pd.read_excel(arquivo_excel, usecols=['CD_SETOR', 'v0001', 'V00001'])
    # Conta o número de linhas
    linhas_excel = len(df_excel)
    print("Linhas presentes: " + str(linhas_excel) + " (Esperado: 468099)")
    
    # Conta setores com código nulo ou vazio
    nulos_excel = df_excel['CD_SETOR'].isnull().sum()
    print("Setores com código vazio/nulo: " + str(nulos_excel) + " (Esperado: 0)")
    
    # Conta setores duplicados
    duplicatas_excel = df_excel['CD_SETOR'].duplicated().sum()
    print("Setores duplicados: " + str(duplicatas_excel) + " (Esperado: 0)")
    
except Exception as e:
    # Exibe mensagem de erro caso não consiga ler o Excel
    print("Erro ao ler o Excel: " + str(e))

# Finaliza a auditoria
print("\nAuditoria Concluida!")
```

# Auditoria de Integridade dos Arquivos
Esta célula realiza uma auditoria de integridade nos arquivos essenciais do projeto, verificando a existência, tamanho, e consistência dos dados em um banco SQLite e um arquivo Excel. Ela reporta possíveis erros e inconsistências, como ausência de arquivos, setores nulos ou duplicados, e quantidade de colunas.


# 1. Importação de bibliotecas
Esta célula importa as bibliotecas necessárias para manipulação de dados, acesso ao banco SQLite e operações com arquivos.

```python
# Importa as bibliotecas necessárias
import pandas as pd
import sqlite3
import os
```

# 2. Mensagem inicial e definição de caminhos
Esta célula exibe a mensagem inicial da auditoria e define os caminhos dos arquivos essenciais.

```python
# Exibe mensagem inicial de auditoria e define caminhos dos arquivos
print("Iniciando a Auditoria de Integridade dos Arquivos...")
caminho_bd = '../banco_de_dados/'
arquivo_sqlite = caminho_bd + 'Banco_IVS_Essencial.db'
arquivo_excel = caminho_bd + 'Base_IVS_Essencial.xlsx'
```

# 3. Verificação da existência dos arquivos
Esta célula verifica se os arquivos SQLite e Excel existem e exibe mensagens de sucesso ou erro.

```python
# Verifica se os arquivos SQLite e Excel existem e exibe mensagens
if os.path.exists(arquivo_sqlite):
    tamanho_mb = os.path.getsize(arquivo_sqlite) / (1024 * 1024)
    print("OK: Banco SQLite encontrado (" + str(round(tamanho_mb, 2)) + " MB)")
else:
    print("ERRO: Banco SQLite não encontrado!")

if os.path.exists(arquivo_excel):
    tamanho_mb = os.path.getsize(arquivo_excel) / (1024 * 1024)
    print("OK: Ficheiro Excel encontrado (" + str(round(tamanho_mb, 2)) + " MB)")
else:
    print("ERRO: Ficheiro Excel não encontrado!")
```

# 4. Auditoria no Banco SQLite
Esta célula realiza a auditoria de integridade no banco SQLite, verificando quantidade de linhas, setores nulos, duplicados e número de colunas.

```python
# Auditoria de integridade no banco SQLite
print("\n--- Auditoria no Banco SQLite ---")
try:
    conexao = sqlite3.connect(arquivo_sqlite)
    cursor = conexao.cursor()
    cursor.execute("SELECT COUNT(*) FROM base_ivs_bruta")
    linhas_sql = cursor.fetchone()[0]
    print("Linhas presentes: " + str(linhas_sql) + " (Esperado: 468099)")
    cursor.execute("SELECT COUNT(*) FROM base_ivs_bruta WHERE CD_SETOR IS NULL OR CD_SETOR = ''")
    nulos_sql = cursor.fetchone()[0]
    print("Setores com código vazio/nulo: " + str(nulos_sql) + " (Esperado: 0)")
    cursor.execute("SELECT COUNT(*) - COUNT(DISTINCT CD_SETOR) FROM base_ivs_bruta")
    duplicatas_sql = cursor.fetchone()[0]
    print("Setores duplicados: " + str(duplicatas_sql) + " (Esperado: 0)")
    df_head = pd.read_sql("SELECT * FROM base_ivs_bruta LIMIT 0", conexao)
    colunas_sql = len(df_head.columns)
    print("Colunas presentes: " + str(colunas_sql) + " (Esperado: 32)")
    conexao.close()
except Exception as e:
    print("Erro ao ler o SQLite: " + str(e))
```

# 5. Auditoria no arquivo Excel
Esta célula realiza a auditoria de integridade no arquivo Excel, verificando quantidade de linhas, setores nulos, duplicados e reportando erros de leitura.

```python
# Auditoria de integridade no arquivo Excel
print("\n--- Auditoria no Ficheiro Excel ---")
print("Lendo o Excel para conferencia (isto pode demorar um minuto)...")
try:
    df_excel = pd.read_excel(arquivo_excel, usecols=['CD_SETOR', 'v0001', 'V00001'])
    linhas_excel = len(df_excel)
    print("Linhas presentes: " + str(linhas_excel) + " (Esperado: 468099)")
    nulos_excel = df_excel['CD_SETOR'].isnull().sum()
    print("Setores com código vazio/nulo: " + str(nulos_excel) + " (Esperado: 0)")
    duplicatas_excel = df_excel['CD_SETOR'].duplicated().sum()
    print("Setores duplicados: " + str(duplicatas_excel) + " (Esperado: 0)")
except Exception as e:
    print("Erro ao ler o Excel: " + str(e))
```

# 6. Finalização da auditoria
Esta célula exibe a mensagem de conclusão da auditoria.

```python
# Exibe mensagem de conclusão da auditoria
print("\nAuditoria Concluida!")
```
